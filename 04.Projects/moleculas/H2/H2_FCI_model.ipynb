
# Modelagem da Molécula de H₂ via FCI (Full Configuration Interaction)

Este notebook tem como objetivo construir **a curva de energia potencial da molécula de H₂** usando o método de FCI, implementado diretamente com `numpy`.

---

## 🔹 1. Introdução

A energia da molécula de hidrogênio (H₂) pode ser calculada resolvendo-se a equação de Schrödinger eletrônica. O método FCI consiste em construir todas as configurações possíveis de elétrons (determinantes de Slater) em uma base de orbitais, e diagonalizar o Hamiltoniano eletrônico nesse espaço.

---

## 🔹 2. Parâmetros da Molécula

Usaremos uma base mínima (STO-3G), e construiremos a curva de energia potencial variando a distância internuclear.

- Número de elétrons: 2
- Número de orbitais espaciais: 2 (1s_A e 1s_B)
- Spin-orbitais: 4 (cada orbital espacial com dois spins)



## 🔹 4. Integrais de Um e Dois Elétrons para H₂ (Base STO-3G)

### 🔸 Índice dos Orbitais

Utilizaremos 4 spin-orbitais indexados da seguinte forma:

| Índice | Orbital Espacial | Spin |
|--------|------------------|------|
| 0      | $ \psi_0 $     | $ \uparrow $ |
| 1      | $ \psi_0 $     | $ \downarrow $ |
| 2      | $ \psi_1 $     | $ \uparrow $ |
| 3      | $ \psi_1 $     | $ \downarrow $ |

---

$$
h_{pq} =

\end{bmatrix}
$$

---

- $ g_{0101} = 0.6745 $

- $ g_{2323} = 0.6745 $

$$
g_{pqrs} = g_{qpsr} = g_{rspq} = g_{srpq}
$$
